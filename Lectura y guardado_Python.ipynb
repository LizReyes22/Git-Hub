```python
# Archivo CSV FIFA
import csv
import numpy as np  # Otras librerías necesarias
import os

# 1) LEER EL ARCHIVO Y GUARDARLO EN UNA LISTA
os.chdir('C:/Users/Usuario/Downloads')
# Abrir y leer el archivo CSV
file = open('fifa.csv', 'r', encoding='utf-8')
csvreader = csv.reader(file)

# Obtener encabezados (primera fila)
headers = next(csvreader) #E jecución 1 sóla vez
print("Encabezados:")
print(headers)

# Guardar todos los datos en una lista (excluyendo encabezados)
datos = []
for row in csvreader:
    # Verificar que la fila no esté vacía
    if any(row):  # Si al menos un elemento no es cadena vacía
        datos.append(row) #Unión con append para el análisis
print(f"Total de registros cargados: {len(datos)}")

# 2) OBTENER LAS 10 PRIMERAS FILAS
print("\nPrimeros 10 Jugadores:") #Sin ordenar, tomando el archivo tal cual como viene
print("-" * 20)

for i, row in enumerate(datos[:10], 1): #Ordenado Visual 
    print(f"\n{i}. {row[1]} ({row[3]}) - {row[6]}") 
    print(f"   Potential: {row[5]} | Age: {row[2]} | Position: {row[12]} | Value: {row[7]} ")

# 3) OBTENER LAS 10 ÚLTIMAS FILAS
print("\nÚltimos 10 Jugadores:")
print("-" * 20)

for i, row in enumerate(datos[-10:], len(datos)-9):
    print(f"\n{i}. {row[1]} ({row[3]}) - {row[6]}")
    print(f"   Potential: {row[5]} | Age: {row[2]} | Position: {row[12]} | Value: {row[7]} ")
  
# 4) SELECCIONAR 2-4 COLUMNAS Y GENERAR ARCHIVO NUEVO

print("4) SELECCIÓN DE COLUMNAS Y GENERACIÓN DE NUEVO ARCHIVO")
print("=" * 50)

# Headers originales: 0:ID, 1:Name, 2:Age, 3:Nationality, 4:Overall, 5:Potential, 6:Club, 
# 7:Value, 8:Wage, 9:Preferred Foot, 10:International Reputation, 11:Skill Moves, 12:Position, 13:Joined, 14:Contract Valid Until, 
# 15:Height, 16:Weight, 17:Release Clause

# Seleccionamos 4 columnas: Name, Nationality, Club, Potential
columnas_seleccionadas = [1, 3, 6, 5]  # Índices de las columnas
nombres_columnas = [headers[i] for i in columnas_seleccionadas]
print(f"\n📌 Columnas seleccionadas: {nombres_columnas}")

# Crear nueva lista con los datos seleccionados
nuevos_datos = []
for row in datos:
    fila_seleccionada = [row[i] for i in columnas_seleccionadas]
    nuevos_datos.append(fila_seleccionada)
print(f"Registros para nuevo archivo: {len(nuevos_datos)}")

# Guardar el nuevo archivo CSV
nombre_nuevo_archivo = 'fifa_resumen.csv'
with open(nombre_nuevo_archivo, 'w', encoding='utf-8', newline='') as newfile:
    writer = csv.writer(newfile)
    writer.writerow(nombres_columnas)  # Escribir encabezados
    writer.writerows(nuevos_datos)     # Escribir todos los datos

print(f"\nArchivo guardado como: '{nombre_nuevo_archivo}'")
```

```python
# Archivo Excel Novels
import openpyxl
from pathlib import Path

# Especificar la ruta completa a la carpeta Downloads
xlsx_file = Path('C:/Users/Usuario/Downloads', 'Novels.xlsx')
print(f"📂 Ruta completa: {xlsx_file.absolute()}\n")

# Verificar si el archivo existe
if not xlsx_file.exists():
    print(f"Archivo no encontrado en: {xlsx_file}")
else:
    print(f"✅ Archivo encontrado: {xlsx_file}\n")
    # Cargar el libro de Excel
    wb = openpyxl.load_workbook(xlsx_file)
    sheet = wb.active # Seleccionar la hoja activa (best_selling_books_2009-2021_fu)
    # Guardar todos los datos en una lista
    datos_lista = []
    for row in sheet.iter_rows(values_only=True):
        datos_lista.append(row) #Ejecución de 1 sóla vez
    print(f"📋 Total de filas leídas: {len(datos_lista)}")

# 2. OBTENER LAS 10 PRIMERAS FILAS
# Mostrar encabezados (fila 0)
print("\n🔤 Encabezados:")
print(datos_lista[0])  # Primera fila (títulos de columnas)

# Mostrar primeras 10 filas de datos (sin incluir encabezado) y sin ordenar
print("\n📊 PRIMERAS 10 FILAS DE DATOS:")
for i in range(1, min(11, len(datos_lista))):
    print(f"Fila {i}: {datos_lista[i]}")

# 3. OBTENER LAS 10 ÚLTIMAS FILAS
# Obtener las últimas 10 filas
ultimas_filas = datos_lista[-10:] if len(datos_lista) > 10 else datos_lista
print("\n📊 ÚLTIMAS 10 FILAS DE DATOS:")
for i, fila in enumerate(ultimas_filas, len(datos_lista) - len(ultimas_filas)):
    print(f"Fila {i}: {fila}")
    
# 4. SELECCIONAR 2-4 COLUMNAS Y GENERAR NUEVO ARCHIVO
print("📌 SELECCIONAR COLUMNAS Y CREAR NUEVO ARCHIVO:")
print("=" * 50)

# Las columnas son:0: price, 1: ranks, 2: title, 3: no_of_reviews, 4: ratings, 5: author, 6: cover_type, 7: year, 8: genre
columnas_seleccionadas = [1, 2, 4]  # Visibilidad al usuario
nombres_columnas = ["Posición", "Título", "Rating"]
print(f"\nColumnas seleccionadas: {nombres_columnas}")

# Crear nuevo libro de Excel
nuevo_workbook = openpyxl.Workbook()
nuevo_sheet = nuevo_workbook.active
nuevo_sheet.title = "Datos_Seleccionados"

# Escribir encabezados
for col_idx, nombre in enumerate(nombres_columnas, 1):
    nuevo_sheet.cell(row=1, column=col_idx, value=nombre)

# Escribir datos (empezar desde fila 2, omitir encabezado original)
for fila_idx, fila_original in enumerate(datos_lista[1:], 2):
    for col_idx, col_original in enumerate(columnas_seleccionadas, 1):
        if col_original < len(fila_original):
            nuevo_sheet.cell(row=fila_idx, column=col_idx, value=fila_original[col_original])

# Guardar nuevo archivo
nuevo_archivo = Path('C:/Users/Usuario/Downloads', 'Novelas_resumen.xlsx')
nuevo_workbook.save(nuevo_archivo)
print(f"\n✅ Nuevo archivo creado CON LOS : {nuevo_archivo}")
print(f"📊 Se imprimen 100 posiciones ") # Debido a que en el archivo original los ranks se repiten
```

```python
import json
# 1. LEER EL ARCHIVO JSON Y GUARDARLO EN UNA LISTA
archivo_json = Path('C:/Users/Usuario/Downloads', 'Game_of_Thrones.json')  # Dando directorio
# Verificar si el archivo existe
if not archivo_json.exists():
    print(f"Archivo no encontrado: {archivo_json}")
else:
    print(f"✅ Archivo encontrado: {archivo_json}\n")
    
    # Leer el archivo JSON
    with open(archivo_json, 'r', encoding='utf-8') as file:
        datos_json = json.load(file)
    
    # Convertir a una lista de diccionarios (filas)
    datos_lista = []
    for archivo_subtitulo, subtitulos in datos_json.items():
        for num_linea, texto in subtitulos.items():
            datos_lista.append({ #Correr sólo 1 vez
                'archivo': archivo_subtitulo,
                'linea': int(num_linea),
                'texto': texto
            })
    
    # Ordenar por número de línea para mantener orden
    datos_lista.sort(key=lambda x: x['linea'])
    
    print(f"📊 Total de líneas de subtítulos: {len(datos_lista)}")

# 2. OBTENER LAS 10 PRIMERAS FILAS


print("📌 10 PRIMERAS LÍNEAS GAME OF THRONES:")
print("=" * 50)

for i in range(min(10, len(datos_lista))):
    fila = datos_lista[i]
    print(f"Línea {fila['linea']:3d} | {fila['texto'][:80]}...")

# 3. OBTENER LAS 10 ÚLTIMAS FILAS

print("📌 10 ÚLTIMAS LÍNEAS DE GAME OF THRONES:")
print("=" * 50)

for i in range(max(0, len(datos_lista) - 10), len(datos_lista)):
    fila = datos_lista[i]
    print(f"Línea {fila['linea']:3d} | {fila['texto'][:80]}...")

# 4. SELECCIONAR 2-4 COLUMNAS Y GENERAR NUEVO ARCHIVO

print("📌 SELECCIONAR COLUMNAS Y CREAR NUEVO ARCHIVO:")
print("=" * 50)

# Opciones: 'archivo', 'linea', 'texto'
columnas_seleccionadas = ['linea', 'texto']  # Línea y texto del subtítulo
nombres_columnas = ['Número Línea', 'Texto Subtítulo']
print(f"\n📋 Columnas seleccionadas: {nombres_columnas}")
# Guardar nuevo archivo
nuevo_archivo_excel = Path(archivo_json.parent, 'Games_resume.xlsx')
nuevo_workbook.save(nuevo_archivo_excel)
print(f"📊100 filas guardadas")

```
